```python
def coeff_E4_E6_j1(E):
    #E est un modèle sous forme réduite de courbe elliptique
    #On reprend les formules de Schoof section 7.
    a = E.a4()
    b = E.a6()
    j = E.j_invariant()
    E4 = (-a)*48
    E6 = (b)*864 #Correction d'une erreur de signe dans Schoof
    j1 = -(E6/E4)*j
    return E4, E6, j1

def coeff_p1(E,El,l):
    #On calcul p1 avec la formule 7.2, en combinant avec le théorème 7.3 et la formule 7.1 iii
    F = E.base_field()
    j = E.j_invariant()
    jl = El.j_invariant()
    Psi = classical_modular_polynomial(l)
    ZXY = Psi.parent()
    (X,Y) = ZXY.gens()
    PsiX = Psi.derivative(X)
    PsiY = Psi.derivative(Y)
    PsiXX = (PsiX.derivative(X))(j,jl)
    PsiXY = (PsiX.derivative(Y))(j,jl)
    PsiYY = (PsiY.derivative(Y))(j,jl)
    
    E4, E6, j1 = coeff_E4_E6_j1(E)
    El4, El6, jl1 = coeff_E4_E6_j1(El)
    
    logdiff = -(j1*j1*PsiXX + 2*l*j1*jl1*PsiXY + l*l*jl1*jl1*PsiYY)/(j1*PsiX(j,jl))

    p1 = F(l/2)*(logdiff + ((E4^2/E6) - l*(El4^2/El6))/2 + ((E6/E4) - l*(El6/El4))*F(2/3))

    return p1

coeff_p1(E,E_phi,l) #Fker_vérifié = x^15 + 39x^14 ... donc -p1/2 = 39 : on doit trouver p1 = 23 : E_phi n'est pas le bon modèle
```

```python
def coeff_courbe_l_isogene(E,l,h):
    F = E.base_field()
    j = E.j_invariant()
    a = E.a4()
    b = E.a6()
    
    Psi = classical_modular_polynomial(l)
    ZXY = Psi.parent()
    (X,Y) = ZXY.gens()
    PsiX = (Psi.derivative(X))(j,h)
    PsiY = (Psi.derivative(Y))(j,h)
    
    h1 = F(-18/l)*(b/a)*(PsiX/PsiY)*j
    A = F(-1/48)*(h1^2)/((h-1728)*h)
    B = F(-1/864)*(h1^3)/((h-1728)*h^2)
    A = A
    B = B
    return A,B

h = Fq(-57)
(A,B)=coeff_courbe_l_isogene(E,31,h)
E_coef = EllipticCurve(Fq,[A,B])
E_coef
```

```python
def polynome_ker(Fwp, exp, l):
    prec = (l - 1)/2 #le nombre de coefficient que l'on cherche
    relation = Fwp - exp(z)
    d = 2
    i = prec - 1
    eval = [aa[i] for i in [0 .. (prec - 1)]] #Cette liste va contenir les valeurs des coefficients du polynômes cherché
    while d <= l-1:
        coef = relation[d]     #On regarde une équation donnée
        eval[i] = 0            #On remplace a_i par 0 pour trouver sa valeur
        coef = -coef(eval)
        coef = Fq(coef)
        eval[i] = coef       #On retient la valeur de a_i trouvée
        i = i - 1            #On passe au coefficient suivant
        d = d + 2            #On passe à la relation suivante
    Fker = eval[0] + z^15
    for i in [1 .. (prec - 1)]:
        Fker = Fker + eval[i]*z^i
    return Fker

p1 = 23
Fker = polynome_ker(Fwp, exp,31)
Fker
```

```python
#Trouver un idéal de norme donner : Pas Cornacchia à priori, juste une racine mod l !
#Cf notes de cours Jean François Biasse

from sage.rings.finite_rings.integer_mod import square_root_mod_prime

def ideal_de_norme(l,O):
    disc = O.discriminant()
    if kronecker(disc,l) == 1:
        disc = Mod(disc,l)
        sq = square_root_mod_prime(disc,l) #Lien avec seysen ? Prendre le min des deux racines vu dans N
        sm = Mod(-sq,l)
        sq = ZZ(sq)
        sm = ZZ(sm)
        sq = min(sm,sq)
        #print(sq)
        return NumberFieldOrderIdeal(O,[l, -sq + f*t])
    elif kronecker(disc,l) == 0:
        return NumberFieldOrderIdeal(O,[l, f*t])
    else:
        raise "l est inerte"

#La racine utilisée ici n'est pas le coefficient b de la forme quadratique

def forme_de_seysen(l,O):

    #On cherche b de manière exhaustive en supposant l petit
    #Mieux avec thm chinois ?
    
    disc = O.discriminant()
    if kronecker(disc,l) == 1:
        b = 0                                         #Partir de 0 ou de 1 ?
        while Mod(b,4*l)^2 != Mod(disc,4*l):
            b = b+1
        ql = BinaryQF([l,b, int((b^2 - disc)/(4*l))])
        return ql

def ideal_de_seysen(l,O):
    ql = forme_de_seysen(l,O)
    return NumberFieldOrderIdeal(O,ql)
        
```

```python
#Algo 2 Jao : trouver une famille génératrice avec des idéaux de normes premiers petits et autorisés

#Seysen partie 3 : N = 2log^2(D) suffirait

def formes_generatrices (D,N,test_norm):
    Idéaux = []
    Formes = []
    Primes = []
    i = 2
    while i < N:
        if kronecker(D,i) == 1 and (test_norm)%i != 0:
            Primes.append(i)
            qi = forme_de_seysen(i,O)
            Li = NumberFieldOrderIdeal(O,qi)
            Idéaux.append(Li)
            Formes.append(qi)
        i = next_prime(i)
    return Idéaux, Formes, Primes

test_norm = fm*p*Cn
Idéaux, Formes, Primes = formes_generatrices(D,30,test_norm) #Trouver la borne : Analyse de Jao
Formes
```

```python
#Algo 3 Jao (Hafney / McCurley) Factoriser L dans la base précédente

def facto_complet(facto,Primes):
    # On utilise le fait que facto et Primes soit triée
    #facto = factor(a)
    k = len(facto)
    if len(Primes) >= k:
        i = 0
        j = 0
        while i < k and j < len(Primes):
            if facto[i][0] == Primes[j]:
                i = i+1
                j = j+1
            else:
                j = j+1
        return i == k
    else:
        return False
            
        

def factorisation(L,D,N,test_norm):
    Idéaux, Formes, Primes = formes_generatrices (D,N,test_norm)
    ql = L.quadratic_form().reduced_form()
    k = len(Primes)
    expx = [0]*k
    a = ql[0]
    facto = factor(a)
    ko = 3
    BORNE_z = k #A préciser avec l'analyse de Jao : = rac(log(D/3))/z
    relation = ql
    while facto_complet(facto,Primes) == False:
        
        # Choix d'exposant au hasard
        while ko >= 3 and ko <= BORNE_z:
            ko = 0
            for i in [0 .. k-1]:
                expi = randint(0,int((N/Primes[1])^2))
                expx[i] = expi
                if expi != 0:
                    ko = ko + 1
        # calcul de l* ( fi^exp(xi) pour tout i ) pour chercher une relation "simplifiable"
        relation = ql
        for i in [0 .. k-1]:
            facteur = power(Formes[i],exp[i]) #ecrire une fonction power efficace : double and square avec reduction à chaque étape
            facteur = facteur.reduced_form()
            relation = ql*facteur
            relation = relation.reduced_form()
        a = relation[0]
        facto = factor(a) #Si a factorise complétement dans Primes, c'est gagné avec Seysen

    #Calcul des valuations de a en chaque Primes, et change le signe selon la méthode Seysen
    expu = []
    ka = len(facto)
    b = relation[1]
    i = 0
    for j in [0 .. k-1]:
        if i < ka :
            if Primes[j] == facto[i][0] :
                expu.append(facto[i][1])
                i = i + 1
                bi = Formes[i][1]
                if Mod( b , 2*Primes[i]) != Mod( bi , 2*Primes[i]):
                    expu[-1] = -expu[-1]
            else :
                expu.append(0)
        else :
            expu.append(0)
                        

    #inverser la relation pour trouver L:
    expe = []
    factoL = []
    for j in [0 .. k-1] :
        ej = expu[j] - expx[j]
        if ej > 0 :
            expe.append(ej)
            factoL.append(Idéaux[j].conjugate())
        elif ej < 0 :
            expe.append(-ej)
            factoL.append(Idéaux[j])

    return expe, factoL
```

```python
from sage.rings.finite_rings.integer_mod import square_root_mod_prime
from sage.rings.finite_rings.integer_mod import square_root_mod_prime_power
from sage.rings.number_field.order_ideal import NumberFieldOrderIdeal
from sage.schemes.elliptic_curves.hom_velusqrt import EllipticCurveHom_velusqrt
from sage.schemes.elliptic_curves.weierstrass_morphism import *
from sage.groups.generic import order_from_multiple
from sage.schemes.elliptic_curves.ell_curve_isogeny import compute_isogeny_bmss
from sage.schemes.elliptic_curves.hom_frobenius import EllipticCurveHom_frobenius
import time

def ideal_de_norme(l,f,D):

    #Trouver un idéal de norme l premier : Cf notes de cours Biasse
    #On se place dans un ordre de discriminant D et de conducteur f
    
    if kronecker(D,l) == 1:
        D = Mod(D,l)
        sq = square_root_mod_prime(D,l) #Lien avec Seysen ? Prendre le min des deux racines vu dans N ? (Optionnel)
        sm = Mod(-sq,l)
        sq = ZZ(sq)
        sm = ZZ(sm)
        sq = min(sm,sq)
        #print(sq)
        return NumberFieldOrderIdeal(O,[l, -sq + f*t]) #Forme donnée dans le cours de Biasse
    elif kronecker(D,l) == 0:
        return NumberFieldOrderIdeal(O,[l, f*t])
    else:
        raise "l est inerte"

#La racine utilisée ici n'est pas le coefficient b de la forme quadratique associée.




def forme_de_norme(l,D):

    #Remarque : Seysen dmande une racine spécifique, la plus petite vu dans N. Mais cela n'est pas nécessaire pour la suite (Cf Cohen)
    
    if kronecker(D,l) == 1:
        if l == 2 :
            d = Mod(D,8)
            b = square_root_mod_prime_power(d,2,3)
            b = ZZ(b)
        else :
            x = Mod(D,4)
            d = Mod(D,l)
            y = square_root_mod_prime(d,l)
            b = x.crt(y)
            b = ZZ(b)
        ql = BinaryQF([l,b, int(int((b^2 - D))/int(4*l))])
        return ql



def base_wk(b,dk):

    #Dans K = Q(t) quadratique de discriminant dk, on se donne b = (x + yt)/2 dans OK
    #On veut écrire b dans la base [1,wk], wk = (dk + t)/2
    
    x = b[0]
    y = b[1]
    wx = x - dk*y
    wy = 2*y
    return (wx,wy)

def base_fq1(b,fm,dk,tracef):
    
    # On représente le froebenius fq par le complexe (tracef + fm*t)/2 dans K = Q(t).
    # Il existe s entier tel que fq = fm*wk + s.
    
    s1 = (-fm*dk + tracef)/2
    (x,y) = base_wk(b,dk)
    fx = x*fm - y*s1
    return (fx,y)

def base_fq2(b,fm,dk,tracef):

    # On représente le froebenius fq par le complexe (tracef - fm*t)/2 dans K = Q(t).
    # Il existe s entier tel que -fq = fm*wk + s.
    # Comme fq et fm*wk ne sont pas égaux à un entier prêt, cela semble être la mauvaise représentation de fq pour la normalisation
    
    s2 = (-fm*dk - tracef)/2
    (x,y) = base_wk(b,dk)
    fx = x*fm - y*s2
    return (fx,-y)


def eval_fq(P,kq,E):

    #calcul Frobenius(P) sur une courbe E définie sur Fq
    #Le point P peut appartenir à un extension, il faut alors donner E telle que P appartienne à E.
    if P == E(0):
        return P
    else :
        frob = EllipticCurveHom_frobenius(E, kq)
        #Q = [P[0],P[1]]
        #Q[0] = (Q[0]).pth_power(kq)
        #Q[1] = (Q[1]).pth_power(kq)
        #Q = E(Q)
        return frob(P)
        
```

```python
# Calculs dans le groupe de classe :


def formes_generatrices (D,N,test_norm): #Trouver la borne : Analyse de Jao

#Algo 2 Jao : Trouver une famille génératrice du groupes de classes, vu comme formes quadratiques réduites,
# En n'utilisant que des idéaux de normes premiers petits splits et autorisés pour le calcul d'isogénies horizontales
# On génére simplement une famille de N idéaux, N supposé suffisament grand pour que leurs classes soient génératrices.
# Seysen partie 3 : N = 2log^2(D) suffirait (Admet Riemann Généralisé)
    
    Idéaux = []
    Formes = []
    Primes = []
    i = 2
    while i < N:
        if kronecker(D,i) == 1 and (test_norm)%i != 0:
            Primes.append(i)
            qi = forme_de_norme(i,D)
            Li = NumberFieldOrderIdeal(O,qi)
            Idéaux.append(Li)
            Formes.append(qi)
        i = next_prime(i)
    return Idéaux, Formes, Primes



def facto_complet(a,Primes):
    
    #On veut savoir si un entier a de factorisation facto = factor(a) ne contient que des premiers de la liste Primes
    # On utilise le fait que facto et Primes soit triée

    test = a
    k = len(Primes)
    for i in [0 .. k-1]:
        pi = Primes[i]
        while test%pi == 0:
            test = test/pi
    return test == 1
            
            
        

def factorisation(L,D,N,Borne_z,test_norm):

    #Algo 3 Jao : (Hafney / McCurley) Factoriser L dans la base précédente
    #On cherche au hasard à ramener la classe de L à une classe simplifiable par le théorème 3.1 de Seysen.
    #On représente les classes par des formes quadratiques réduites : c'est nécessaire pour se ramener à Seysen
    #Borne_z est une borne donnée par Jao dans l'analyse de l'algorithme. Le nombres de facteurs de L ne dépassera pas Borne_z
    
    Idéaux, Formes, Primes = formes_generatrices (D,N,test_norm)
    print('liste des nombres premiers autorisées')
    print(Primes)
    ql = (L.quadratic_form()).reduced_form() #Représente la classe de L
    print(ql)
    k = len(Primes)
    
    a = ql[0]
    
    relation = ql        # Partant de ql, on va chercher une classe simplifiable
    
    expx = [0]*k
    liste_indice = []
    print('Recherche de relation')
    compte = 0
    while facto_complet(a,Primes) == False:
        compte = compte + 1
        # Choix d'exposant au hasard, on veut entre 3 et Borne_z coefficient non nul (Cf algo 3 Jao)
        # Autoriser 1 ou 2 comme nombre d'exposant possible ?
        
        expx = [0]*k       # La liste des exposants notée xi par Jao
        nb_indice = randint(3,Borne_z) 
        liste_indice = []    # on choisit les idéaux qui vont intervenir dans la factorisation, au hasard.
        while len(liste_indice) < min(nb_indice,k):
            i = randint(0,k-1)
            if i not in liste_indice :
                liste_indice.append(i)
        
        
        for i in liste_indice:
            expi = randint(1,int((N/Primes[i])^2))   #On attribue à chaque indice un exposant xi, bornée d'après Jao.
            expx[i] = expi

        
        # calcul de ql*( fi^exp(xi) pour tout i ) pour chercher une relation "simplifiable"
        relation = ql
        for i in [0 .. k-1]:
            for _ in [1 .. (expx[i])]:
                relation = (relation*Formes[i]).reduced_form()
        a = relation[0] #Si a factorise complétement dans Primes, c'est gagné avec Seysen
        
    facto = factor(a) 
    print('relation simplifiable trouvée')
    print(relation)
    print('nombre de tentatives :', compte)
    print('exposants associés')
    print(Primes)
    print(expx)
    print('Factorisation de a')
    print(facto)
    
    
    #Calcul des valuations de a en chaque Primes, et change le signe selon la méthode Seysen
    expu = [0]*k     #liste des exposant ui de Jao
    ka = len(facto)
    b = relation[1]
    j = 0
    for i in [0 .. k-1]:
        if j < ka :
            if Primes[i] == facto[j][0] :  
                expu[i] = facto[j][1]
                bi = Formes[i][1]
                if Mod( b , 2*Primes[i]) != Mod( bi , 2*Primes[i]):
                    expu[i] = -expu[i]
                j = j + 1
        
    
    #inverser la relation pour trouver L:
    
    expe = []     #liste des exposants ei de Jao
    factoL = []
    for i in [0 .. k-1] :
        ei = expu[i] - expx[i]
        if ei > 0 :
            expe.append(ei)
            factoL.append(Idéaux[i])
        elif ei < 0 :
            expe.append(-ei)
            factoL.append(Idéaux[i].conjugate())
            
    
    return expe, factoL #, compte
```

```python
# Données sur les noyaux :

def deg_extension_noyau(l,L,f,fm,dk,tracef): #On calcul le degré d'extension r necessaire pour obtenir la L-torsion
    
    #On écrit L dans la base [1,fq]
    ql = L.quadratic_form()
    cible = (-ql[1] + f*t)/2                    #Deuxieme élément d'une base sous forme standard .
    (c,d) = base_fq1(cible, fm, dk, tracef)     # a priori d = 1
    #(c2,d2) = base_fq2(cible, fm, dk, tracef) Inutile sauf si fq2 peut être la bonne représentation du frobenius.

    #On calcul la taille de l'extension : elle ne dépend que de c et l
    Fl.<xl> = GF(l)
    b = GF(l)(-c)
    r = order_from_multiple(b, l - 1, operation='*')
    return r


def card_extension_noyau(E,l,L,f,fm,dk,tracef):       #On calcul tout ce qui concerne l'extension dans laquelle chercher le noyau de l-isogénie.
    print('début calculs extensions noyaux')
    
    #On écrit L dans la base [1,fq]
    ql = L.quadratic_form()
    cible = (-ql[1] + f*t)/2                    #Deuxieme élément d'une base sous forme standard .
    (c,d) = base_fq1(cible, fm, dk, tracef)     # a priori d = 1
    #(c2,d2) = base_fq2(cible, fm, dk, tracef) Inutile sauf si fq2 peut être la bonne représentation du frobenius.

    #On calcul la taille de l'extension : elle ne dépend que de c et l
    Fl.<xl> = GF(l)
    b = GF(l)(-c)
    r = order_from_multiple(b, l - 1, operation='*')
    q_ker = q^r
    print('taille extension :',r)

    #On calcul le cardinal de la courbe. Théorème de tate : toutes les courbes isogènes ont même cardinal
    print('debut calcul C_ker')
    Fker.<xk> = GF(q_ker) 
    E_ker = EllipticCurve(Fker, [E.a4(),E.a6()]) 
    C_ker = E_ker.cardinality()    #Le déduire de celui de E ?
    if C_ker%l != 0 :              #Par construction C_ker doit être divisible par l
        raise 'Erreur Pas de l torsion'
    print('C_ker ok :', C_ker)

    #On calcul les n1 possibles tels que E_ker soit le groupe (Z/n1 Z) x (Z/n2 Z) avec n1|n2. 
    #On utilise n1^2 | C_ker et q_ker = 1 mod n1
    print('début calcul des n1 possibles')
    carrés = facteurs_carrés(C_ker,q_ker)
    k = len(carrés)
    print('Nb de n1 à tester :',k)
    print('Fin calculs noyau extension')
    return q_ker, c, C_ker, carrés
```

```python
#Construction du groupe de torsion E[L], méthode 1



#Pour trouver E[L], besoin de c et d : deux choix selon fq1 ou fq2 à priori, on va tout de même privilégier le choix 1.

def noyaux(c,d,P,Q,kq,E):
    
    #Q est dans E[L] ssi dans E[l] et dans E[c+dfq] (voir rapport, on utilise le fait que fm et l sont premier entre eux)
    #On teste pour chaque P +iQ générateur des sous groupes de E[l] d'ordre l. 
    
    for i in [0 .. l-1]:
        Pi = P + i*Q
        cPi = c*Pi
        fPi = eval_fq(Pi,kq,E)
        dfPi = d*fPi
        pPi = cPi + dfPi
        if pPi == E(0,1,0):
            return Pi
    cQ = c*Q
    fQ = eval_fq(Q,kq,E)
    dfQ = d*fQ
    pQ = cQ + dfQ
    if pQ == E(0,1,0):
        return Q
    else:
        return None

def groupes_torsions(E,L,l,kq,f,fm,dk,tracef):

    #On veut trouver un générateur de E[L].
    #On mets en arguments des données du problèmes précalculées.
    
    #On écrit L dans la base [1,fq]
    ql = L.quadratic_form()
    cible = (-ql[1] + f*t)/2                    #Deuxieme élément d'une base sous forme standard .
    (c1,d1) = base_fq1(cible, fm, dk, tracef)
    (c2,d2) = base_fq2(cible, fm, dk, tracef) #Inutile sauf si fq2 peut être la bonne représentation du frobenius.

    #On trouve des générateurs de E[l]
    Fl = E.division_field(l)          #ETAPE TROP LENTE, le degrée de l'extension est de l'odre de l d'après l'article de Broker.
    El = EllipticCurve(Fl, [E.a4(),E.a6()])
    G1, G2 = El.torsion_basis(l)

    #On en déduit des générateurs de E[L] selon le choix de fq1 ou fq2. Dans la suite on ne conserve que fq1.
    return El, noyaux(c1,d1,G1,G2,kq,El), noyaux(c2,d2,G1,G2,kq,El)



#Fin méthode 1
```

```python
#Construction du groupe de torsion E[L], méthode 2

def facteurs_carrés(n,qr):

    #On cherche la liste de tout les entiers n1 tels que n1^2 divise n et qr = 1 mod n1
    
    nmax = gcd(n,qr-1)
    carrés = []
    n1 = 2
    while n1 <= int(sqrt(nmax)):
        if (nmax%n1 == 0 and n%(n1^2) == 0):
            carrés.append(n1)
        n1 = n1 + 1
    if n%(nmax^2) == 0:
        carrés.append(nmax)
    return carrés



def generateur_noyau(E,l,L,q,kq,f,fm,dk,tracef,q_ker, c, C_ker, carrés):
    print('début test générateur')
    Fker.<xk> = GF(q_ker) 
    E_ker = EllipticCurve(Fker, [E.a4(),E.a6()]) 
    k = len(carrés)
    test = False
    essais = 0
    while test == False :
        essais = essais + 1
        P = E_ker.random_point()
        K1 = (C_ker/l)*P
        if (K1 != E_ker(0)) :
            test = (eval_fq(K1,kq,E_ker) == (-c)*K1)
        elif P != E_ker(0) :
            for i in [0 .. k-1] :
                n1 = carrés[i]
                n2 = C_ker/n1     #On suppose P d'ordre n2
                if n2%l == 0 :
                    K1 = (n2/l)*P    #lent ???
                    if (K1 != E_ker(0)) :
                        if (eval_fq(K1,kq,E_ker) == (-c)*K1):
                            print('Nb essais generateur du noyau :')
                            print(essais)
                            K1.set_order(l)    #On evite à sage de recalculer l'ordre du point
                            return E_ker, K1
            test = False
        else :
            test = False
            
    print('Nb essais generateur du noyau :')
    print(essais)
    K1.set_order(l) #On evite à sage de recalculer l'ordre du point
    return E_ker, K1

#Fin méthode 2
```

```python
#Construction du groupe de torsion E[L], méthode 3 : Recherche d'un polynôme qui décrit le noyau

def coeff_courbe_l_isogene(E,l,h,Psi_l):
    F = E.base_field()
    j = E.j_invariant()
    a = E.a4()
    b = E.a6()
    
    ZXY = Psi_l.parent()
    (X,Y) = ZXY.gens()
    PsiX = (Psi_l.derivative(X))(j,h)
    PsiY = (Psi_l.derivative(Y))(j,h)
    
    h1 = F(-18/l)*(b/a)*(PsiX/PsiY)*j
    A = F(-1/48)*(h1^2)/((h-1728)*h)
    B = F(-1/864)*(h1^3)/((h-1728)*h^2)
    A = A*(F(l)^4) #Normalisation
    B = B*(F(l)^6)
    return A,B



def test_noyau(E,Fker,L,f,fm,dk,tracef):    #ERREUR ICI A CORRIGER
    
    a = E.a4()
    b = E.a6()
    F = E.base_field()
    PF.<X,Y>=PolynomialRing(F, order='lex')
    FE = Y^2 - X^3 - a*X - b
    Fker = Fker(X)
    IdQuo = PF.ideal([Fker, FE])
    Quo_test = PF.quotient(IdQuo)

    FrobX = Quo_test(X)^q
    FrobY = Quo_test(Y)^q

    ql = L.quadratic_form()
    cible = (-ql[1] + f*t)/2                    #Deuxieme élément d'une base sous forme standard .
    (c,d) = base_fq1(cible, fm, dk, tracef)
    val = Mod(-c,l)
    val = ZZ(val)
    if abs(val) > abs(val - l):
        val = val - l
    
    phi = E.scalar_multiplication(val)
    Fcx, Fcy = phi.rational_maps()

    
    FcxD = Fcx.denominator()
    FcxN = Fcx.numerator()
    FcxD = Quo_test(FcxD(X,Y))
    FcxN = Quo_test(FcxN(X,Y))
    Test_X = FcxN/FcxD

    
    FcyD = Fcy.denominator()
    FcyN = Fcy.numerator()
    FcyD = Quo_test(FcyD(X,Y))
    FcyN = Quo_test(FcyN(X,Y))
    Test_Y = FcyN/FcyD

    return Test_X == FrobX and Test_Y == FrobY



def polynome_ker(E,l,L,f,fm,dk,tracef,j,Psi_ni):
    
    PFq.<z> = PolynomialRing(Fq)
    Psi = Psi_ni(j,z)
    test_j_invariant = (gcd(Psi,z^q - z)).factor()
    h = -(test_j_invariant[0][0][0])
    (A,B)=coeff_courbe_l_isogene(E,l,h,Psi_ni)
    E_coef = EllipticCurve(Fq,[A,B])
    Fker = compute_isogeny_bmss(E, E_coef, l) #On suppose p > 4l + 4 
    if test_noyau(E,Fker,L,f,fm,dk,tracef):
        return Fker
    else:
        h = -(test_j_invariant[1][0][0])
        (A,B)=coeff_courbe_l_isogene(E,l,h,Psi_ni)
        E_coef = EllipticCurve(Fq,[A,B])
        Fker = compute_isogeny_bmss(E, E_coef, l)
        if test_noyau(E,Fker,L,f,fm,dk,tracef):
            return Fker
        else:
            raise 'Aucun j trouvé'
```

```python
#Gestion de l'idéal principal :

def Cornacchia_avec_racine(n,d,s): #Inutile ?

    #On veut trouver un élément (x + y*f*t)/2 dans l'idéal, de norme n, à partir d'une racine s du discrimant D, s déjà obtenue.
    #Jao et Hardy : Cornacchia général : on résoud 4*n = x²+y²|D| ( Avec les notations de Hardy, f = 1, g = |D|, m = D )
    #Il faut à priori que 4*n et D soit premier entre eux ... ne semble pas garanti. 
    print('Condition pour Cornacchia :')
    print(gcd(4*n,abs(d)))
    
    #On applique l'étape 6 de l'algorithme d'Hardy, c'est à dire l'étape de division euclidienne de Cornacchia
    print('Suite des restes :')
    print(s)
    r1 = n
    while s > int(sqrt(4*n)) :
        r2 = r1%s
        r1 = s
        s = r2
        print(s)
    x = s
    y = (4*n - s^2)/abs(d)
    print( 'Valeurs de y^2 et y :')
    print(y)
    print(sqrt(y))
    if y.is_square():
        y = int(sqrt(y))
        return (x,y)
    else :
        raise 'Pas de carré fin Cornacchia'

def calcul_de_alpha(L,Idéaux,Exposant,D):
    B = L     # B sera l'idéal principal dont on cherche un générateur.
    k = len(Idéaux)
    m = 1    #ici m désigne le produit des normes des idéaux premier qui factorise L. alpha = beta/m
    for i in [0 .. k-1] :
        pi = Idéaux[i]
        for _ in range (Exposant[i]):
            B = B*(pi.conjugate())
            m = m*(pi.norm())

    if B.is_principal():    #Remarque : B est primitif comme produit d'idéaux primitif
        beta = (B.gens_reduced())[0]    #Par les formes quadratiques, via une solution pour représenter 1. Coder soit même
        bO = NumberFieldOrderIdeal(O,beta)
        if bO == B:
            print('résultat de réduction correct')
            return beta, m
        else:
            print('résultat de réduction conjugué')
            beta = (x - y*t)/2
            return beta, m
    else:
        print('Erreur de factorisation')
        raise 'Erreur de factorisation'



def eval_phic(P,E_eval,Composantes):
    #On calcul l'image d'un point Q par la suite de composantes de phi.
    #Q appartient à E_eval, donc potentiellement pas à E(Fq). 
    dépard = E_eval
    for phi in Composantes:
        P = dépard(P)
        P = phi(P)
        dépard = phi.codomain()
        dépard = EllipticCurve(GF(q^n), [dépard.a4(),dépard.a6()])
    return P

def isom_normalisation(Ec,ua,va,m,fm):
    #On calcul l'isomorphisme qui corrige l'a courbe d'arrivée, pour normaliser phi. 
    #Il suffit de corriger l'action de [alpha] car les composantes sont normalisées.
    # alpha = (ua + va fq )/(fm*m)
    u = Fp(ua)*((Fp(m*fm))^-1)
    isom = WeierstrassIsomorphism(Ec, (u^(-1), 0, 0, 0))
    return isom

def endo_alpha(Q,En,kq,u,v,m,fm,Cn):
    #On veut calculer l'image d'un point Q par l'endomorphisme alpha. Q appartient à la courbe d'arrivée des composantes, normalisée. 
    #alpha = (u + v fq)/(mfm)
    denom = (Mod(m,Cn)*Mod(fm,Cn))^(-1) #inversible par choix des nombres premiers depuis le début. 
    c = Mod(u,Cn)*denom
    d = Mod(v,Cn)*denom
    cQ = c*Q      #a bien un sens car c est un entier
    fqQ = eval_fq(Q,kq,En)
    dfqQ = d*fqQ
    alphaQ = En(cQ) + En(dfqQ) # ici correction d'un erreur de type inconnue, en précisant que l'on veut additionner des points de En
    return alphaQ
```

```python
def Broker(E,p,q,kq,E_eval,P,n,K,O,l,N,Borne_z):

    début = time.time()
    #Calcul des données du problème :
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk
    
    CE = E.cardinality_pari()
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = sqrt(Dm/(K.discriminant()))
    
    Cn = E_eval.cardinality()
    test_norm = fm*p*Cn   #On souhaite éviter tout nombre premier qui divise test_norm
    
    #Application des algos précédents :
    
    L = ideal_de_norme(l,f,D)
    
    print('début factorisation')
    Exposant, Idéaux = factorisation(L,D,N,Borne_z,test_norm)
    print('fin factorisation')
    print('temps :')
    print(time.time() - début)
    
    print('début Composantes')
    Composantes = composantes_de_phi(E_eval,q,n,kq,Exposant,Idéaux,f,fm, dk ,tracef)
    print('fin Composantes')
    print('temps :')
    print(time.time() - début)
    
    beta, m = calcul_de_alpha(L,Idéaux,Exposant,D)
    u, v = base_fq1(beta , fm , dk, tracef)

    if len(Composantes) > 0:
        Ec = (Composantes[-1]).codomain()
    else:
        Ec = E_eval
    Pc = eval_phic(P,E_eval,Composantes)

    isom = isom_normalisation(Ec,u,v,m,fm)
    Pc = isom(Pc)
    E_phi = isom.codomain()

    Pc = endo_alpha(Pc, E_phi,kq,u,v,m,fm,Cn)
    print(time.time() - début)
    
    return Pc, E_phi
    





def Broker_BMSS(E,p,q,kq,E_eval,P,n,K,O,l,N,Borne_z):

    début = time.time()
    #Calcul des données du problème :
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk
    
    CE = E.cardinality_pari()
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = sqrt(Dm/(K.discriminant()))
    
    Cn = E_eval.cardinality()
    test_norm = fm*p*Cn   #On souhaite éviter tout nombre premier qui divise test_norm
    
    #Application des algos précédents :
    
    L = ideal_de_norme(l,f,D)
    
    print('début factorisation')
    Exposant, Idéaux = factorisation(L,D,N,Borne_z,test_norm)
    print('fin factorisation')
    print('temps :')
    print(time.time() - début)
    
    print('début Composantes')
    Composantes = composantes_de_phi_polynome(E_eval,q,n,kq,Exposant,Idéaux,f,fm, dk ,tracef)
    print('fin Composantes')
    print('temps :')
    print(time.time() - début)
    
    beta, m = calcul_de_alpha(L,Idéaux,Exposant,D)
    u, v = base_fq1(beta , fm , dk, tracef)

    if len(Composantes) > 0:
        Ec = (Composantes[-1]).codomain()
    else:
        Ec = E_eval
    Pc = eval_phic(P,E_eval,Composantes)

    isom = isom_normalisation(Ec,u,v,m,fm)
    Pc = isom(Pc)
    E_phi = isom.codomain()

    Pc = endo_alpha(Pc, E_phi,kq,u,v,m,fm,Cn)
    print(time.time() - début)
    return Pc, E_phi




def sqrtvelu(E,q,kq,E_eval,K,O,l):

    q = p^kq
    début = time.time()
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = Dm/(K.discriminant())
    
    L = ideal_de_norme(l,f,D)
    q_ker,c,C_ker,carrés = card_extension_noyau(E,l,L,f,fm,dk,tracef)
    E_ker, K1 = generateur_noyau(E,l,L,q,kq,f,fm,dk,tracef,q_ker,c,C_ker,carrés)
    phi = phi_from_ker(K1,l,E_ker)
    print('temps :')
    print(time.time() - début)
    
    return phi
```

```python

def factorisation(L,D,N,Borne_z,test_norm):

    #Algo 3 Jao : (Hafney / McCurley) Factoriser L dans la base précédente
    #On cherche au hasard à ramener la classe de L à une classe simplifiable par le théorème 3.1 de Seysen.
    #On représente les classes par des formes quadratiques réduites : c'est nécessaire pour se ramener à Seysen
    #Borne_z est une borne donnée par Jao dans l'analyse de l'algorithme. Le nombres de facteurs de L ne dépassera pas Borne_z
    
    Idéaux, Formes, Primes = formes_generatrices (D,N,test_norm)
    ql = (L.quadratic_form()).reduced_form() #Représente la classe de L
    k = len(Primes)
    
    a = ql[0]
    
    relation = ql        # Partant de ql, on va chercher une classe simplifiable
    
    expx = [0]*k
    liste_indice = []
    compte = 0
    while facto_complet(a,Primes) == False:
        compte = compte + 1
        # Choix d'exposant au hasard, on veut entre 3 et Borne_z coefficient non nul (Cf algo 3 Jao)
        # Autoriser 1 ou 2 comme nombre d'exposant possible ?
        
        expx = [0]*k       # La liste des exposants notée xi par Jao
        nb_indice = randint(3,Borne_z) 
        liste_indice = []    # on choisit les idéaux qui vont intervenir dans la factorisation, au hasard.
        while len(liste_indice) < min(nb_indice,k):
            i = randint(0,k-1)
            if i not in liste_indice :
                liste_indice.append(i)
        
        
        for i in liste_indice:
            expi = randint(1,int((N/Primes[i])^2))   #On attribue à chaque indice un exposant xi, bornée d'après Jao.
            expx[i] = expi

        
        # calcul de ql*( fi^exp(xi) pour tout i ) pour chercher une relation "simplifiable"
        relation = ql
        for i in [0 .. k-1]:
            for _ in [1 .. (expx[i])]:
                relation = (relation*Formes[i]).reduced_form()
        a = relation[0] #Si a factorise complétement dans Primes, c'est gagné avec Seysen
        
    facto = factor(a) 
    
    
    #Calcul des valuations de a en chaque Primes, et change le signe selon la méthode Seysen
    expu = [0]*k     #liste des exposant ui de Jao
    ka = len(facto)
    b = relation[1]
    j = 0
    for i in [0 .. k-1]:
        if j < ka :
            if Primes[i] == facto[j][0] :  
                expu[i] = facto[j][1]
                bi = Formes[i][1]
                if Mod( b , 2*Primes[i]) != Mod( bi , 2*Primes[i]):
                    expu[i] = -expu[i]
                j = j + 1
        
    
    #inverser la relation pour trouver L:
    
    expe = []     #liste des exposants ei de Jao
    factoL = []
    for i in [0 .. k-1] :
        ei = expu[i] - expx[i]
        if ei > 0 :
            expe.append(ei)
            factoL.append(Idéaux[i])
        elif ei < 0 :
            expe.append(-ei)
            factoL.append(Idéaux[i].conjugate())
            
    
    return expe, factoL , compte
```

```python
def test_extensions_noyaux(E,K,O,N,E_eval):
    
    #Calcul des données du problème :
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk

    CE = E.cardinality_pari()
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = sqrt(Dm/(K.discriminant()))

    Cn = E_eval.cardinality()
    test_norm = fm*p*Cn 

    Idéaux, Formes, Primes = formes_generatrices (D,N,test_norm)
    nb = len(Idéaux)
    extensions = []
    for i in range(nb):
        #On écrit L dans la base [1,fq]
        L = Idéaux[i]
        l = Primes[i]
        ql = L.quadratic_form()
        cible = (-ql[1] + f*t)/2                    #Deuxieme élément d'une base sous forme standard .
        (c,d) = base_fq1(cible, fm, dk, tracef)     # a priori d = 1
        #(c2,d2) = base_fq2(cible, fm, dk, tracef) Inutile sauf si fq2 peut être la bonne représentation du frobenius.

        #On calcul la taille de l'extension : elle ne dépend que de c et l
        Fl.<xl> = GF(l)
        b = GF(l)(-c)
        r = order_from_multiple(b, l - 1, operation='*')
        extensions.append((l,r))
    
    return extensions
```

```python

#Version adaptée au calcul de noyau


def formes_generatrices_adaptée (D,f,fm,dk,tracef,N,test_norm):

    #On reprend l'algorithme de Jao en évitant la redondance des classes, et en gardant le meilleur représentant pour le calcul d'extension
    
    Classes = [] #liste des formes quadratiques réduites, sans doublons
    Représentants = [] #liste d'idéaux représentant chaque classe
    Normes = [] #Liste des normes des représentants
    Extensions = [] #liste des degrées d'extensions de leurs L torsions
    Primes = [] #liste triée des nombres premiers dont on a vu la classe au moins une fois
    Idéaux = [] #Liste de leurs idéaux de Seysen
    Formes = [] #Liste de leurs formes de Seysen
    
    i = 2
    while i < N:
        if kronecker(D,i) == 1 and (test_norm)%i != 0:
            Primes.append(i)
            #Données de l'idéal de Seysen
            qi = forme_de_seysen(i,D)
            cli = qi.reduced_form()
            Li = NumberFieldOrderIdeal(O,qi)
            ri = deg_extension_noyau(i,Li,f,fm,dk,tracef)
            
            Idéaux.append(Li)
            Formes.append(qi)
            
            #Données de son conjugué
            qib = BinaryQF([qi[0],-qi[1],qi[2]])
            clib = qi.reduced_form()
            Lib = NumberFieldOrderIdeal(O,qib)
            rib = deg_extension_noyau(i,Lib,f,fm,dk,tracef)

            nouveaux = True
            if len(Classes) == 0 :
                nouveaux = False
                Classes.append(cli)
                Représentants.append(Li)
                Normes.append(i)
                Extensions.append(ri)
            
            #test si Li est un représentant interessant
            h = len(Classes)
            for c in [0 .. h-1]:
                if (cli == Classes[c]) and (i^ri < ((Normes[c])^(Extensions[c]))) :
                    nouveaux = False
                    Normes[c] = i
                    Représentants[c] = Li
                    Extensions[c] = ri
            if nouveaux :
                Classes.append(cli)
                Représentants.append(Li)
                Normes.append(i)
                Extensions.append(ri)   

            if clib != cli :
            #test si Lib est un représentant interessant
            
                nouveaux = True
                h = len(Classes)
                for c in [0 .. h-1]:
                    if (clib == Classes[c]) and (i^rib < ((Normes[c])^(Extensions[c]))) :
                        nouveaux = False
                        Normes[c] = i
                        Représentants[c] = Lib
                        Extensions[c] = rib
                if nouveaux :
                    Classes.append(clib)
                    Représentants.append(Lib)
                    Normes.append(i)
                    Extensions.append(rib)
        i = next_prime(i)

    # On cherche encore des représentant, sans chercher de nouvelles classes
    h = len(Classes)

    while i < 2*N :
        if kronecker(D,i) == 1 and (test_norm)%i != 0:
            
            #Données de l'idéal de Seysen
            qi = forme_de_seysen(i,D)
            cli = qi.reduced_form()
            Li = NumberFieldOrderIdeal(O,qi)
            ri = deg_extension_noyau(i,Li,f,fm,dk,tracef)
            
            #Données de son conjugué
            qib = BinaryQF([qi[0],-qi[1],qi[2]])
            clib = qi.reduced_form()
            Lib = NumberFieldOrderIdeal(O,qib)
            rib = deg_extension_noyau(i,Lib,f,fm,dk,tracef)

            for c in [0 .. h-1]:
                if (cli == Classes[c]) and (i^ri < ((Normes[c])^(Extensions[c]))) :
                    Normes[c] = i
                    Représentants[c] = Li
                    Extensions[c] = ri
                if (clib == Classes[c]) and (i^rib < ((Normes[c])^(Extensions[c]))) :
                    Normes[c] = i
                    Représentants[c] = Lib
                    Extensions[c] = rib

        i = next_prime(i)
        
    return Classes, Représentants, Normes, Extensions, Primes, Idéaux, Formes

def factorisation_adaptée(L,D,f,fm,dk,tracef,N,Borne_z,test_norm):

    #On reprend tel quel l'algo de factorisation :
    
    Classes, Représentants, Normes, Extensions, Primes, Idéaux, Formes = formes_generatrices_adaptée (D,f,fm,dk,tracef,N,test_norm)
    print('liste des nombres premiers autorisées')
    print(Primes)
    ql = (L.quadratic_form()).reduced_form() #Représente la classe de L
    print(ql)
    k = len(Primes)
    
    a = ql[0]
    
    relation = ql        # Partant de ql, on va chercher une classe simplifiable
    
    expx = [0]*k
    liste_indice = []
    print('Recherche de relation')
    while facto_complet(a,Primes) == False:
        
        # Choix d'exposant au hasard, on veut entre 3 et Borne_z coefficient non nul (Cf algo 3 Jao)
        # Autoriser 1 ou 2 comme nombre d'exposant possible ?
        
        expx = [0]*k       # La liste des exposants notée xi par Jao
        nb_indice = randint(3,Borne_z) 
        liste_indice = []    # on choisit les idéaux qui vont intervenir dans la factorisation, au hasard.
        while len(liste_indice) < min(nb_indice,k):
            i = randint(0,k-1)
            if i not in liste_indice :
                liste_indice.append(i)
        
        
        for i in [0 .. k-1]:
            if i in liste_indice:
                expi = randint(1,int((N/Primes[1])^2))   #On attribue à chaque indice un exposant xi, bornée d'après Jao.
                expx[i] = expi

        
        # calcul de ql*( fi^exp(xi) pour tout i ) pour chercher une relation "simplifiable"
        relation = ql
        for i in [0 .. k-1]:
            for _ in [1 .. (expx[i])]:
                relation = (relation*Formes[i]).reduced_form()
        a = relation[0] #Si a factorise complétement dans Primes, c'est gagné avec Seysen
        
    facto = factor(a) 
    print('relation simplifiable trouvée')
    print(relation)
    print('exposants associés')
    print(Primes)
    print(expx)
    print('Factorisation de a')
    print(facto)
    
    
    #Calcul des valuations de a en chaque Primes, et change le signe selon la méthode Seysen
    expu = [0]*k     #liste des exposant ui de Jao
    ka = len(facto)
    b = relation[1]
    j = 0
    for i in [0 .. k-1]:
        if j < ka :
            if Primes[i] == facto[j][0] :  
                expu[i] = facto[j][1]
                bi = Formes[i][1]
                if Mod( b , 2*Primes[i]) != Mod( bi , 2*Primes[i]):
                    expu[i] = -expu[i]
                j = j + 1
        
    
    #inverser la relation pour trouver L:
    
    expe = []     #liste des exposants ei de Jao
    factoL = []
    for i in [0 .. k-1] :
        ei = expu[i] - expx[i]
        if ei > 0 :
            expe.append(ei)
            factoL.append(Idéaux[i])
        elif ei < 0 :
            expe.append(-ei)
            factoL.append(Idéaux[i].conjugate())

    #expe, factol factorisation à adapter : On change de représentant pour chaque classe, pour notre préféré.

    h = len(Classes)
    k = len(factoL)
    expf = [0]*h

    print(Classes)
    for i in [0 .. k-1] :
        exp = expe[i]
        Pi = factoL[i]
        qi = Pi.quadratic_form()
        Cli = qi.reduced_form()   #Cli est l'une classe que l'on a vu pendant Forme_generatrice_adaptée
        for c in [0 .. h-1]:
            if Cli == Classes[c] :
                expf[c] = expf[c] + exp
    

    print('Factorisation trouvée :')
    print(Normes)
    print(expf)
        
    return expf, Représentants, Normes, Classes




def composantes_de_phi_adapté(E,q,kq,Exposant,Idéaux,Classes,f,fm,dk,tracef):

    #Algo 4 de Jao (Algo de Broker) : On déduit de la factorisation de L une suite d'isogénies.
    
    dépard = E   #Le point de dépard sera en réalité E_eval car on souhaite évaluer un point Q. 
    k = len(Idéaux)
    composantes = []
    for i in [0 .. k-1]:
        pi = Idéaux[i]
        exp = Exposant[i]
        if exp != 0:
            print('Composante de degré :')
            print(pi.norm())
            print('Exposant :')
            print(exp)
            q_ker, c ,C_ker,carrés = card_extension_noyau(E,pi.norm(),pi,f,fm,dk,tracef)
            for i in [1 .. exp]:
                print('calcul composante numéro', i)
                phic = phi_from_L(dépard,pi,pi.norm(),q,kq,f,fm,dk,tracef,q_ker,c,C_ker,carrés)
                composantes.append(phic)
                dépard = phic.codomain()
                dépard = EllipticCurve(Fq, [dépard.a4(),dépard.a6()]) #On se replace sur le corps de base pour éviter une escalade d'extensions
    return composantes


    
def calcul_de_alpha_adapté(L,Idéaux,Exposant,D):
    B = L     # B sera l'idéal principal dont on cherche un générateur.
    h = len(Idéaux)
    m = 1    #ici m désigne le produit des normes des idéaux premier qui factorise L. alpha = beta/m
    for c in [0 .. h-1] :
        rep = Idéaux[c]
        exp = Exposant[c]
        if exp> 0:
            for _ in [1 .. exp]:
                m = m*(rep.norm())
                B = B*(rep.conjugate())

    if B.is_principal():    #A FAIRE : gérer le cas ù B n'est pas primitif
        beta = (B.gens_reduced())[0]    #Par les formes quadratiques, via une solution pour représenter 1. Coder soit même
        bO = NumberFieldOrderIdeal(O,beta)
        if bO == B:
            print('résultat de Cornacchia correct')
            return beta, m
        else:
            print('résultat de Cornacchia conjugué')
            beta = (x - y*t)/2
            return beta, m
    else:
        print('Erreur de factorisation')
        raise 'Erreur de factorisation'

```

```python
def Broker_adapté(E,p,q,kq,E_eval,Q,n,K,O,l,N,Borne_z):

    début = time.time()
    #Calcul des données du problème :
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk
    
    CE = E.cardinality_pari()
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = sqrt(Dm/(K.discriminant()))
    
    Cn = E_eval.cardinality()
    test_norm = fm*p*Cn   #On souhaite éviter tout nombre premier qui divise test_norm
    
    #Application des algos précédents :
    
    L = ideal_de_norme(l,f,D)
    
    print('début factorisation')
    Exposant, Idéaux, Normes, Classes = factorisation_adaptée(L,D,f,fm,dk,tracef,N,Borne_z,test_norm)
    print('fin factorisation')
    print('temps :')
    print(time.time() - début)
    print(Normes)
    print(Exposant)
    
    print('début Composantes')
    Composantes = composantes_de_phi_adapté(E_eval,q,kq,Exposant,Idéaux,Classes,f,fm, dk ,tracef)
    print('fin Composantes')
    print('temps :')
    print(time.time() - début)
    
    beta, m = calcul_de_alpha_adapté(L,Idéaux,Exposant,D)
    u, v = base_fq1(beta , fm , dk, tracef)

    if len(Composantes) > 0:
        Ec = Composantes[-1].codomain()
    else:
        Ec = E_eval
    Qc = eval_phic(Q,Composantes)

    isom = isom_normalisation(Ec,u,v,m,fm)
    Qc = isom(Qc)
    E_phi = isom.codomain()

    Qc = endo_alpha(Qc, E_phi,kq,u,v,m,fm,Cn)

    return Qc, E_phi
```

```python
def calcul_alpha_reduction(L,Rep,D):
    Repb = Rep.conjugate()
    B = L*Repb
    gen = B.gens_reduced()
    if len(gen) == 1 :
        beta = gen[0]
    else:
        print('Erreur de reduction')
        raise 'Erreur de reduction'
    return beta

calcul_alpha_reduction(L,Rep,D)
```

```python
def sqrtvelu_avec_reduction(E,q,kq,P,E_eval,K,O,l):

    q = p^kq
    début = time.time()
    dk = K.discriminant()
    f = O.conductor()
    D = f^2*dk
    tracef = E.trace_of_frobenius()
    Dm = tracef^2 - 4*q
    fm = Dm/(K.discriminant())
    Cn = E_eval.cardinality()
    test_norm = fm*p*Cn 
    
    L = ideal_de_norme(l,f,D)
    
    Rep = représentant_réduit(L)
    m = Rep.norm()

    if kronecker(D,m) == 1 and (test_norm)%m != 0 :
        m = Rep.norm()
    else :
        print('Besoin de factoriser')
        raise 'Besoin de factoriser'
    
    q_ker,c,C_ker,carrés = card_extension_noyau(E,m,Rep,f,fm,dk,tracef)
    E_ker, K1 = generateur_noyau(E,m,Rep,q,kq,f,fm,dk,tracef,q_ker,c,C_ker,carrés)
    phi_rep = phi_from_ker(K1,l,E_ker)

    
    beta = calcul_alpha_reduction(L,Rep,D)
    u, v = base_fq1(beta , fm , dk, tracef)

    Ec = phi_rep.codomain()
    Pc = phi_rep(P)

    isom = isom_normalisation(Ec,u,v,m,fm)
    Pc = isom(Pc)
    E_phi = isom.codomain()

    Pc = endo_alpha(Pc, E_phi,kq,u,v,m,fm,Cn)
    print(time.time() - début)
    print('temps :')
    return Pc, E_phi
```

```python
def base_wk(b,dk):

    #Dans K = Q(t) quadratique de discriminant dk, on se donne b = (x + yt)/2 dans OK
    #On veut écrire b dans la base [1,wk], wk = (dk + t)/2
    
    x = b[0]
    y = b[1]
    wx = x - dk*y
    wy = 2*y
    return (wx,wy)

def base_fq1(b,fm,dk,tracef):
    
    # On représente le froebenius fq par le complexe (tracef + fm*t)/2 dans K = Q(t).
    # Il existe s entier tel que fq = fm*wk + s.
    
    s1 = (-fm*dk + tracef)/2
    (x,y) = base_wk(b,dk)
    fx = x*fm - y*s1
    return (fx,y)

base_fq1(ll,fm,dk,tracef) # permet d'évaluer l'endomorphisme associé, si fm premier avec cardinal
```

```python
def deg_extension_torsion(E,l,Borne_extension):
    
    # D'après l'article MMSTV proposition 2
    
    Fl = GF(l)
    
    if kronecker(l,D) == 1:
        qL = forme_de_norme(l,D)
        gen = (-qL[1] + f*t)/2
        (c,d) = base_fq1(gen, fm, dk, tracef)
        val1 = Fl(-c)
        r1 = order_from_multiple(val1, l - 1, operation='*')
        val2 = val1/(Fl(q))
        r2 = order_from_multiple(val2, l - 1, operation='*')
        return lcm(r1,r2)
        
    elif kronecker(l,D) == 0:
        L = ideal_de_norme(l,f,D)
        qL = L.quadratic_form()
        gen = (-qL[1] + f*t)/2
        (c,d) = base_fq1(gen, fm, dk, tracef)
        val = Fl(-c)
        r = order_from_multiple(val, l - 1, operation='*')
        test = order_from_multiple(Fl(q), l - 1, operation='*')
        if test%2 == 0:
            return 2*r
        else:
            return r
            
    else:         #Cas inerte non résolu
        r = 1
        E_test = E
        test = False
        while r < Borne_extension and test == False:
            T = E_test.abelian_group().torsion_subgroup(l)
            test = T.invariants() == (l,l)
            r = r+1
            E_test = E.change_ring(GF(q^r))
        return r
```

```python
def liste_premier_acceptable(E, Borne_premier, Borne_extension):
    # Renvoie une liste de petit nombre premier de torsion accessible.
    acceptables = []
    extensions = []
    i = 2
    while i < Borne_premier:
        r = deg_extension_torsion(E,i,Borne_extension)
        if r < Borne_extension:
            acceptables.append(i)
            extensions.append(r)
        i = next_prime(i)
    return acceptables, extensions

acceptables, extensions = liste_premier_acceptable(E, 14, 7)
acceptables, extensions
```

```python
def facto_complet_sans_puissance(a,Primes):
    
    #On veut savoir si un entier a de factorisation facto = factor(a) ne contient que des premiers de la liste Primes
    # On utilise le fait que facto et Primes soit triée

    test = a
    k = len(Primes)
    for i in [0 .. k-1]:
        pi = Primes[i]
        if test%pi == 0:
            test = test/pi
    return test == 1



def ajustement_N_dim124(Nb, Nc, liste, Borne_coef):

    #On cherche des entiers m et n tel que N = Nb*m + Nc*n soit acceptable. 
    
    test = False
    while test == False :
        m = randint(1,Borne_coef)
        n = randint(1,Borne_coef)
        N = Nb*m + Nc*n
        test = facto_complet_sans_puissance(N,liste)
    return m,n

m, n = ajustement_N_dim124(Nb, Nc, acceptables, 5)
N = m*Nb + n*Nc
N, N.factor(), m, n
```
